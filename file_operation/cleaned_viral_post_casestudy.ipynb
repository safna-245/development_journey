```python
#Create a summary showing Platform, Number of Posts, Total Views, Total Likes, Total Shares, Total Comments, and Average Views.

platforms = [p.get("Platform") for p in viral_posts]

summary = {}

for platform in platforms:

    posts = [p for p in viral_posts if p.get("Platform") == platform]

    views = [int(p.get("Views")) for p in posts]

    likes = [int(p.get("Likes")) for p in posts]

    shares = [int(p.get("Shares")) for p in posts]

    comments = [int(p.get("Comments")) for p in posts]

    summary[platform] = {
        "Number of Posts": len(posts),

        "Total Views": sum(views),

        "Total Likes": sum(likes),

        "Total Shares": sum(shares),

        "Total Comments": sum(comments),
        
        "Average Views": sum(views) / len(views)
    }

summary

```
