movie case study

```python
movies = [
    ["kgf","kannada",150,2005,8],
    ["balan","malayalam",130,2026,7],
    ["ramayanam","hindi",150,2026,8.5],
    ["abcd","malayalam",160,2008,6],
    ["goatlife","malayalam",150,2024,9],

]
```

```python
#all movie titles


all_movie_titles = {m[0] for m in movies}
all_movie_titles
```

```python
all_movie_language = [m[1] for m in movies]

all_movie_language
```

```python
movie_upper_case  = {m[0].upper() for m in movies}

movie_upper_case
```

```python
#movie titles starts with a

name_filter = [m[0] for m in movies if m[0].startswith("a")]

name_filter
```

```python
#display malayalam movies

malayalam_movies = [m[0] for m in movies if m[1]=="malayalam"]

malayalam_movies
```

```python
#language wise count
languages = [m[1] for m in movies ]

lc = {l:languages.count(l) for l in languages}

lc
```

```python
import matplotlib.pyplot as plt

plt.bar(list(lc.keys()),lc.values())
```

```python
import matplotlib.pyplot as plt

x = lc.keys()

y = lc.values()

plt.bar(x,y)
```

```python
highest_rating = [m[4] for m in movies ]

max(highest_rating)
```

```python
lowest_rating = [m[4] for m in movies]

min(lowest_rating)
```

```python
avg_rating = sum([m[4] for m in movies]) /len(movies) 

avg_rating
```

```python
highest_duration = [m[2] for m in movies ]

max(highest_duration)
```

```python
lowest_duration = [m[2] for m in movies ]

min(lowest_duration)
```

```python
avg_duration= sum([m[2] for m in movies]) /len(movies) 

avg_duration
```

```python
year = [m[3] for m in movies]

y_count = {yc:year.count(yc) for yc in year}

y_count
```

```python
import matplotlib.pyplot as plt

x = y_count.keys()

y = y_count.values()

plt.bar(x,y)
```
