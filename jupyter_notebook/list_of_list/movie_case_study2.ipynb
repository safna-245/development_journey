```python
movies_data = [
    ["dangal", "hindi", 161, 2016, 8.3],       # 0
    ["premam", "malayalam", 156, 2015, 8.3],   # 1
    ["inception", "english", 148, 2010, 8.8],  # 2
    ["vikram", "tamil", 175, 2022, 8.3],       # 3
    ["parasite", "korean", 132, 2019, 8.5],    # 4
    ["rrr", "telugu", 187, 2022, 7.8],         # 5
    ["drishyam", "malayalam", 160, 2013, 8.3], # 6
    ["avatar", "english", 162, 2009, 7.9],     # 7
    ["kumbalangi nights", "malayalam", 135, 2019, 8.5], # 8
    ["sholay", "hindi", 204, 1975, 8.1]        # 9
]

```

```python
#Q1. Find by Language: Print the titles of all Malayalam movies.
  
mal_movie = [m[0] for m in movies_data if m[1] == "malayalam"]

mal_movie
```

```python
#Q2. Find by Year: Print the titles of all movies released after the year 2015.
  
after_2015_movie = [m[0] for m in movies_data if m[3] >2015]

after_2015_movie
```

```python
# ratinh8.5 or higher

count = len([movie[0] for movie in movies_data if movie[4] >= 8.5])

count
```

```python
#Average Duration: Calculate the average duration (in minutes) of all the movies in the dataset.



avg_duration =sum([m[2] for m in movies_data])/len(movies_data)

avg_duration
```

```python
#  Q5. Multiple Conditions: Print the titles of movies that are longer than 150 minutes AND have a rating greater than 8.0.
movie_title = [m[0] for m in movies_data  if m[2] > 150 and m[4] > 8.0]

movie_title
```

```python
#Highest Rating: Find and print the entire record for the movie with the highest rating.

highest_rating = [m[4] for m in movies_data]

highest = max(highest_rating)

highest_rating_movie = [m for m in movies_data if m[4] == highest]

highest_rating_movie

```

```python
#Q7. Oldest Movie: Find the title and release year of the oldest movie in the dataset.
old_movie = [m[3] for m in movies_data]

oldest = min(old_movie)

oldest_movie = [[m[0],m[3]] for m in movies_data if m[3]==oldest]

oldest_movie


```
