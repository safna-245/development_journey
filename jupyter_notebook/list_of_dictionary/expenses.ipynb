```python
owner_summary = {}

for e in expenses:

    owner = e.get("owner")

    amount = e.get("amount")

    if owner in owner_summary:

        owner_summary[owner] += amount

    else:

        owner_summary[owner] = amount

owner_summary
```

```python
x = owner_summary.keys()

y = owner_summary.values()

plt.bar(x,y)
```

```python
#dispaly costly transaction

max_transaction_amount= max([e.get("amount") for e in expenses])


max_transaction = [e for e in expenses if e.get("amount") == max_transaction_amount]

max_transaction
```

```python
#dispaly lowest transaction

min_transaction_amount= min([e.get("amount") for e in expenses])


min_transaction = [e for e in expenses if e.get("amount") == min_transaction_amount]

min_transaction
```
