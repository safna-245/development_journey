```python
#57. Highest average amount by payer.
payer_amount = {}
payer_count = {}

for t in transactions:
    payer = t["payer"]
    amount = t["amount"]

    if payer in payer_amount:
        payer_amount[payer] += amount
        payer_count[payer] += 1
    else:
        payer_amount[payer] = amount
        payer_count[payer] = 1

average = {}

for p in payer_amount:
    average[p] = payer_amount[p] / payer_count[p]


highest = max(average.values())

highest_payer =[p for p in average if average[p] == highest]

highest_payer

    

        
```
