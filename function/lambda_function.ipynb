```python
#adding two variable
add_num = lambda n1,n2: n1+n2

print(add_num(2,3))
```

```python
#subtracting two  variable

sub_num = lambda n1,n2: n1-n2

print(sub_num(30,20))
```

```python
#max_of_two number

max_two = lambda n1,n2:n1 if n1 >n2 else n2

print(max_two(11,13))
```

```python
#lambda function for min of two number

min_two = lambda n1,n2:n1 if n1 <n2 else n2

print(min_two(11,13))
```

```python
#person with highest experience
user_experience = [

    ["hari",6],
    ["abin",1],
    ["bijoy",4],
    ["cibin",2],
    ["akash",7]
]

print(max(user_experience,key = lambda lst:lst[1]))

print(min(user_experience,key = lambda lst:lst[1]))
```

```python
weather =[

    {"district":"tsr","temp":38},
    {"district":"tvm","temp":36},
    {"district":"ekm","temp":40},
    {"district":"pkd","temp":41},
    {"district":"kty","temp":32},

]
#highest temp district
print(max(weather,key=lambda dictionary:dictionary.get("temp")))
#lowest temp  district
print(min(weather,key=lambda dictionary:dictionary.get("temp")))
```
